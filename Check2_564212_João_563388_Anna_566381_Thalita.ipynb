## Fase 1 — Entendimento do Negócio (CRISP-DM)

Esta seção responde às três perguntas propostas pelo enunciado para a primeira fase do CRISP-DM: o que o grupo define como anomalia, qual o impacto de resolvê-la e qual é o problema exato que o modelo resolverá. Nenhuma implementação é realizada nesta etapa; trata-se exclusivamente da formalização do problema.

### 1.1 Contexto e problema de negócio

Plataformas de avaliação de filmes, como o IMDB, dependem de textos escritos por usuários para orientar a escolha de outros espectadores. Esse mecanismo é vulnerável a conteúdos que não refletem uma opinião genuína sobre a obra: propaganda disfarçada de resenha, texto repetitivo gerado em massa e avaliações escritas com o propósito explícito de inflar ou deprimir a reputação de um título.

O problema de negócio é **identificar, entre as avaliações textuais de uma plataforma de filmes, aquelas que apresentam padrões linguísticos suspeitos, e explicar ao analista quais palavras ou trechos motivaram a suspeita**. O sistema não pretende substituir a moderação humana, mas priorizar casos para revisão e tornar a decisão do modelo auditável.

### 1.2 Importância para uma plataforma de filmes

Detectar avaliações suspeitas é relevante pelos seguintes motivos:

* **Credibilidade da plataforma:** avaliações pouco confiáveis reduzem a confiança do público no sistema como um todo.
* **Qualidade da informação:** notas e resenhas influenciam decisões de consumo e a percepção pública das obras; conteúdo manipulado distorce essa referência.
* **Integridade de recomendações e rankings:** sistemas que usam avaliações como entrada herdam o viés introduzido por conteúdo artificial.
* **Eficiência da moderação:** a revisão manual de grandes volumes de texto é cara; um classificador pode ordenar os casos por suspeita e reduzir o esforço humano.
* **Transparência:** a explicação por palavra permite que o moderador valide ou conteste a decisão, o que reduz o risco de penalizar usuários legítimos por falsos positivos.

### 1.3 Objetivos

**Objetivo geral.** Desenvolver um classificador baseado em Rede Neural Recorrente (LSTM) capaz de identificar avaliações de filmes com padrões textuais anômalos e de explicar visualmente, em nível de palavra, os termos que mais contribuíram para essa classificação.

**Objetivos específicos.**

1. Definir de forma operacional o que o grupo considera uma anomalia textual neste contexto.
2. Construir, a partir da base do IMDB, um conjunto de dados rotulado em que parte das avaliações recebe anomalias sintéticas segundo padrões previamente definidos.
3. Treinar um classificador LSTM binário e controlar o *overfitting*.
4. Avaliar o modelo com métricas adequadas a classes possivelmente desbalanceadas (*precision*, *recall*, *F1-score* e curva ROC).
5. Gerar explicações por palavra com LIME e apresentá-las em uma visualização colorida.
6. Verificar criticamente se os termos destacados fazem sentido para um leitor humano, documentando casos de acerto e de falha.
7. Disponibilizar uma interface simples em Gradio para análise de textos inseridos pelo usuário.

### 1.4 Definição do problema de classificação

O problema é formulado como uma **classificação binária supervisionada de textos**:

| Classe | Significado | Descrição |
| :---: | :--- | :--- |
| **0** | Avaliação presumidamente normal | Avaliação original do IMDB, sem alteração. É considerada normal por presunção: a base não oferece garantia de que seja livre de spam real. |
| **1** | Avaliação com anomalia sintética | Avaliação original modificada pelo grupo para conter um dos padrões anômalos definidos na seção 1.5. |

O alvo do modelo é a **presença de um padrão anômalo**, e não a opinião do autor sobre o filme.

### 1.5 Padrões de anomalia sintética

Inicialmente, o grupo adota três padrões. Eles foram escolhidos por serem reconhecíveis em plataformas reais de avaliação e por poderem ser descritos de forma objetiva. A sua forma exata de geração será definida nas fases seguintes.

| Padrão | Descrição | Exemplo ilustrativo |
| :--- | :--- | :--- |
| **Spam promocional** | Inserção de conteúdo publicitário ou de chamada à ação sem relação com o filme, como convites para acessar sites, ofertas ou canais de download. | *"... great movie, visit www.free-movies-now.example and get 50% off today!"* |
| **Repetição excessiva** | Reiteração anormal de palavras, expressões ou frases, característica de texto gerado em massa ou de preenchimento artificial. | *"best best best movie movie movie ever ever ever ..."* |
| **Linguagem de manipulação de avaliações** | Expressões que indicam a intenção de influenciar a nota ou a reputação do título, em vez de descrever uma experiência de assistir ao filme. | *"Give this film 10 stars, everyone must vote 10 to push its rating up."* |

Os exemplos acima são apenas ilustrativos e não representam avaliações reais da base.

### 1.6 Sentimento do IMDB não é spam

A base do IMDB é classicamente rotulada por **sentimento** (positivo ou negativo). Esse rótulo mede a opinião do autor sobre o filme e **não** informa se a avaliação é spam, repetitiva ou manipulativa. Em particular:

* uma crítica fortemente negativa ou fortemente positiva pode ser perfeitamente legítima;
* uma avaliação anômala pode ser positiva ou negativa;
* portanto, os rótulos originais de sentimento **não serão usados como rótulos de anomalia**, e o modelo não deve ser interpretado como um analisador de sentimento.

Na construção do conjunto de dados, o grupo deverá garantir que a presença de anomalia não fique correlacionada com o sentimento original, para que o classificador não aprenda esse atalho.

### 1.7 Limitações da abordagem com anomalias sintéticas

* **Ausência de verdade de campo:** os rótulos da classe 1 são produzidos pelo próprio grupo. O desempenho medido indica a capacidade de reconhecer os padrões injetados, e não de detectar fraude real.
* **Rótulos da classe 0 são presumidos:** a base original pode conter spam real que será tratado como normal, o que introduz ruído nos rótulos.
* **Risco de aprendizado de artefatos:** se os padrões forem muito regulares, o modelo pode memorizar modelos de texto (*templates*) em vez de generalizar.
* **Cobertura restrita:** apenas três padrões são considerados. Spam real é mais variado e adaptativo, e o modelo pode falhar diante de estratégias não previstas.
* **Distribuição artificial das classes:** a proporção entre as classes no conjunto de treino é uma decisão do grupo e provavelmente difere da frequência de anomalias em produção.
* **Explicações aproximadas:** o LIME gera explicações locais por aproximação e pode apresentar instabilidade entre execuções.
* **Alcance das conclusões:** os resultados **não comprovam** a existência ou a detecção de fraude real; servem como prova de conceito de um pipeline de classificação e explicação.

### 1.8 Solução planejada

* **Classificação:** uma rede LSTM supervisionada, treinada para distinguir avaliações presumidamente normais (classe 0) de avaliações com anomalia sintética (classe 1).
* **Explicabilidade:** o LIME será aplicado às predições do modelo para atribuir importância a cada palavra, permitindo uma visualização com destaque em cores.
* **Demonstração:** uma interface em Gradio, em que o usuário insere um texto, solicita a análise e recebe o veredito junto ao texto com as palavras destacadas.

O desenvolvimento seguirá as seis fases do CRISP-DM, de forma incremental, nas seções subsequentes deste notebook.

### 2.1 Fonte e carregamento

O enunciado se refere à "base de dados de reviews do IMDB", sem indicar uma fonte ou biblioteca específica. Foi adotado o **Large Movie Review Dataset v1.0** (Maas et al., 2011), distribuído por Stanford em `aclImdb_v1.tar.gz`, que é a origem original do corpus. Alternativas consideradas:

O arquivo é baixado com `keras.utils.get_file`, que guarda uma cópia em cache e evita novo download, e lido diretamente do `.tar.gz` com `tarfile`, sem extrair os 100 mil arquivos. São usadas apenas as partições rotuladas (`train` e `test`); a pasta `unsup` (avaliações sem rótulo) não é utilizada. O nome do arquivo de cada avaliação (`id_nota.txt`) fornece a nota original, de 1 a 10.

````python
import hashlib
import os
import re
import tarfile
from collections import Counter

import keras
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS

plt.rcParams.update({"figure.figsize": (7, 4), "axes.grid": True, "grid.alpha": 0.3,
                     "axes.spines.top": False, "axes.spines.right": False})
pd.options.display.float_format = "{:,.2f}".format
````

````python
URL = "https://ai.stanford.edu/~amaas/data/sentiment/aclImdb_v1.tar.gz"
ARQUIVO = keras.utils.get_file("aclImdb_v1.tar.gz", URL)

with open(ARQUIVO, "rb") as f:
    sha256 = hashlib.sha256(f.read()).hexdigest()
print(f"Fonte:   {URL}")
print(f"Arquivo: {ARQUIVO} ({os.path.getsize(ARQUIVO) / 1e6:.1f} MB)")
print(f"SHA-256: {sha256}")

registros = []
with tarfile.open(ARQUIVO, "r:gz") as tar:
    for membro in tar:
        partes = membro.name.split("/")
        if membro.isfile() and len(partes) == 4 and partes[1] in ("train", "test") and partes[2] in ("pos", "neg"):
            id_review, nota = partes[3][:-4].split("_")
            texto = tar.extractfile(membro).read().decode("utf-8")
            registros.append((partes[1], "positivo" if partes[2] == "pos" else "negativo", int(id_review), int(nota), texto))

df = pd.DataFrame(registros, columns=["particao", "sentimento", "id_review", "nota", "texto"])
HASH_TEXTOS = pd.util.hash_pandas_object(df["texto"], index=False).sum()  # para conferir, ao final, que os textos não mudaram
df.head(3)
````

### 2.3 Análise dos textos

O comprimento é medido em caracteres (sobre o texto original) e em palavras. Para contar palavras e montar o vocabulário é usada uma tokenização exploratória simples: minúsculas, remoção da etiqueta HTML `<br />` e extração de sequências alfanuméricas, mantendo contrações como *don't*. Ela não substitui a tokenização definitiva da Fase 3.

````python
TOKEN = re.compile(r"[a-z0-9]+(?:'[a-z]+)?")
BR = re.compile(r"<br\s*/?>")


def tokenizar(texto):
    """Tokenização apenas exploratória; a definitiva será feita na Fase 3."""
    return TOKEN.findall(BR.sub(" ", texto.lower()))


contagens = {chave: Counter() for chave in df.groupby(["particao", "sentimento"]).groups}
n_palavras, rep_consec = [], []
for particao, sentimento, texto in zip(df["particao"], df["sentimento"], df["texto"]):
    tokens = tokenizar(texto)
    n_palavras.append(len(tokens))
    rep_consec.append(sum(a == b for a, b in zip(tokens, tokens[1:])))
    contagens[(particao, sentimento)].update(tokens)

df["n_palavras"] = n_palavras
df["n_caracteres"] = df["texto"].str.len()
df["rep_consec"] = rep_consec  # pares de palavras idênticas consecutivas
````

````python
percentis = [0.05, 0.25, 0.5, 0.75, 0.9, 0.95, 0.99]
print("Comprimento das avaliações (palavras e caracteres):")
display(df[["n_palavras", "n_caracteres"]].describe(percentiles=percentis).T)

print("Palavras por sentimento:")
display(df.groupby("sentimento")["n_palavras"].describe(percentiles=percentis))

print(f"Correlação palavras x caracteres: {df['n_palavras'].corr(df['n_caracteres']):.3f}\n")
cortes = [100, 128, 200, 256, 300, 400, 500, 600]
print("Fração das avaliações que cabem inteiras em L palavras:")
pd.Series({L: (df["n_palavras"] <= L).mean() for L in cortes}, name="fração").to_frame().T
````

````python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))
p99 = df["n_palavras"].quantile(0.99)
ax1.hist(df["n_palavras"].clip(upper=p99), bins=60, color="#4f81bd")
ax1.axvline(df["n_palavras"].median(), color="k", linestyle="--", label=f"mediana = {df['n_palavras'].median():.0f}")
ax1.axvline(df["n_palavras"].quantile(0.95), color="#c0504d", linestyle="--", label=f"P95 = {df['n_palavras'].quantile(0.95):.0f}")
ax1.set_title("Distribuição do comprimento das avaliações")
ax1.set_xlabel(f"Palavras (valores acima do P99 = {p99:.0f} agrupados na última barra)")
ax1.set_ylabel("Quantidade de avaliações")
ax1.legend()

grupos = [df.loc[df["sentimento"] == s, "n_palavras"] for s in ("negativo", "positivo")]
ax2.boxplot(grupos, showfliers=False)
ax2.set_xticks([1, 2])
ax2.set_xticklabels(["negativo", "positivo"])
ax2.set_title("Comprimento por sentimento (sem outliers)")
ax2.set_xlabel("Sentimento")
ax2.set_ylabel("Palavras")
plt.tight_layout()
plt.show()
````

* Em caracteres, a mediana é 970 e a média 1.309; a correlação com a contagem de palavras é 0,997, então as duas medidas são redundantes para os fins deste projeto.

## Fase 3 — Preparação dos Dados (CRISP-DM) — Etapa 03: dataset supervisionado

Esta etapa constrói o dataset rotulado que alimentará a LSTM: classe **0** (avaliação original presumidamente normal) e classe **1** (avaliação modificada com anomalia sintética). A prioridade é a confiabilidade: nenhum texto idêntico ou derivado da mesma origem pode aparecer em partições diferentes, e nenhuma regra é ajustada com dados de teste.

Ficam para a etapa seguinte a tokenização definitiva, o *padding* e a conversão em números. Nenhum modelo é treinado aqui.

| Decisão | Escolha |
| :--- | :--- |
| Duplicatas | Cópias equivalentes removidas (1 representante); quase-duplicatas mantidas, mas agrupadas; registros do teste em conflito com o treino são excluídos |
| Partições | Nenhuma avaliação do teste original vai para treino ou validação; validação tirada do treino original; um subconjunto do teste original reservado para generalização |
| Proporção das classes | 50% / 50% em cada partição e sentimento |
| Seleção das anomalias | Sorteio com semente fixa (42), estratificado por partição e sentimento; não usa nota, sentimento nem comprimento |
| Posição da anomalia | Variável, sempre terminando até a 200ª palavra |
| Texto do classificador | Texto original limpo (`<br />` removido, espaços normalizados) mais o trecho inserido |
| Teste de generalização | 1.200 avaliações com anomalias de mecanismos inéditos, fora do treinamento |

### 3.1 Tratamento das duplicatas

**Política** (definida antes de qualquer modelo):

1. **Cópias equivalentes** são textos iguais após uma normalização básica (remoção de `<br />`, minúsculas, remoção de pontuação e espaços repetidos). Mantém-se **um representante** por grupo, dando prioridade ao treino original; as demais cópias são removidas e registradas em uma tabela com o representante correspondente. Se um grupo tivesse sentimentos distintos, seria descartado por inteiro.
2. **Quase-duplicatas** são textos diferentes que compartilham os mesmos 20 primeiros ou os mesmos 20 últimos tokens normalizados (reedições com pequenas correções ou assinaturas do mesmo autor). Elas **não são removidas**, mas recebem um `grupo_id` e **nunca são separadas** entre partições. Se um grupo reúne avaliações do treino original e do teste original, **preserva-se o treino e excluem-se os registros do teste**; eles nunca são transferidos para o treino. Cada exclusão é listada com os registros do treino a que se liga.
3. **Vínculo com a origem:** `id_origem` reproduz o caminho do arquivo original (`partição/sentimento/id_nota`) e é mantido em todo o dataset.

A célula de configuração define a semente e as funções de limpeza e de normalização, e as seguintes investigam as duplicatas.

````python
import random

from scipy.stats import chi2_contingency
from sklearn.metrics import roc_auc_score

SEMENTE = 42
FRACAO_VALIDACAO = 0.15   # parcela da partição original de treino reservada à validação
N_GENERALIZACAO = 1_200   # avaliações da partição original de teste reservadas ao teste de generalização
JANELA_SEGURA = 200       # palavras iniciais que o classificador deverá enxergar (ver 3.3)
POS_MAX = 160             # maior posição de inserção, em palavras
SEG_MAX = 40              # maior trecho inserido, em palavras
TIPOS = ["spam_promocional", "repeticao_excessiva", "manipulacao_avaliacoes"]


def limpar(texto):
    """Limpeza mínima: remove <br /> e normaliza espaços; mantém caixa e pontuação."""
    return re.sub(r"\s+", " ", BR.sub(" ", texto)).strip()


def chave_normalizada(texto):
    """Chave de equivalência: limpeza, minúsculas e remoção de pontuação."""
    return re.sub(r"[\W_]+", " ", limpar(texto).lower()).strip()
````

````python
base = pd.DataFrame({
    "id_origem": df["particao"] + "/" + df["sentimento"].str[:3] + "/" + df["id_review"].astype(str) + "_" + df["nota"].astype(str),
    "particao_original": df["particao"],
    "sentimento": df["sentimento"],
    "texto_original": df["texto"],
})
base["chave"] = base["texto_original"].map(chave_normalizada)


def resumo_duplicatas(chave):
    dup = base.loc[chave.duplicated(keep=False)].assign(k=chave)
    g = dup.groupby("k")
    return {"cópias excedentes": int(chave.duplicated().sum()), "grupos de textos equivalentes": g.ngroups,
            "grupos entre treino e teste": int((g["particao_original"].nunique() > 1).sum()),
            "grupos com sentimentos distintos": int((g["sentimento"].nunique() > 1).sum())}


niveis = {"1. texto exato": base["texto_original"],
          "2. limpeza + minúsculas": base["texto_original"].map(lambda t: limpar(t).lower()),
          "3. chave normalizada (sem pontuação)": base["chave"]}
pd.DataFrame({nome: resumo_duplicatas(chave) for nome, chave in niveis.items()})
````

````python
tokens = base["chave"].str.split()
base["chave_inicio"] = tokens.map(lambda t: " ".join(t[:20]) if len(t) >= 20 else None)
base["chave_fim"] = tokens.map(lambda t: " ".join(t[-20:]) if len(t) >= 20 else None)

unicos = base.drop_duplicates("chave")  # quase-duplicatas além das equivalentes já tratadas
linhas = {}
for coluna in ("chave_inicio", "chave_fim"):
    d = unicos.dropna(subset=[coluna])
    d = d[d.duplicated(coluna, keep=False)]
    g = d.groupby(coluna)
    linhas[coluna] = {"grupos de quase-duplicatas": g.ngroups, "avaliações envolvidas": len(d),
                      "grupos entre treino e teste": int((g["particao_original"].nunique() > 1).sum())}
display(pd.DataFrame(linhas))

d = unicos.dropna(subset=["chave_inicio"])
exemplo = d[d.duplicated("chave_inicio", keep=False)].groupby("chave_inicio").head(2).head(2)
for _, linha in exemplo.iterrows():
    print(f"{linha['id_origem']}: ...{linha['chave'][-90:]}")
````

**Interpretação da investigação.**

* A normalização encontra poucas duplicatas além das exatas: 418 cópias excedentes no texto exato, 422 após limpeza e minúsculas e **424** com a chave normalizada (412 grupos). Em todos os níveis, **123 grupos** cruzam treino e teste e **nenhum** tem sentimentos distintos.
* Há ainda 34 grupos de quase-duplicatas pelo início do texto (71 avaliações; 5 cruzam treino e teste) e 31 pelo final (66 avaliações; 3 cruzam). No exemplo exibido, duas avaliações compartilham o início e terminam de forma diferente. Na amostra inspecionada, os grupos por início eram em geral a mesma avaliação com pequenas edições ou continuações, e os grupos por final eram em geral assinaturas ou trechos finais copiados. Por isso o critério de agrupamento é conservador.

````python
base["prioridade"] = (base["particao_original"] == "test").astype(int)  # em empate, o treino original é mantido
base = base.sort_values(["prioridade", "id_origem"]).reset_index(drop=True)

por_chave = base.groupby("chave")
conflito = por_chave["sentimento"].transform("nunique") > 1     # política: descartar o grupo inteiro (não ocorre)
eh_copia = base.duplicated("chave")
base["n_copias_removidas"] = por_chave["id_origem"].transform("size") - 1
removidas = base.loc[eh_copia, ["id_origem", "particao_original", "sentimento"]].assign(
    id_representante=por_chave["id_origem"].transform("first")[eh_copia],
    particao_representante=por_chave["particao_original"].transform("first")[eh_copia])
n_conflito = int(conflito.sum())
base = base.loc[~eh_copia & ~conflito].reset_index(drop=True)

# quase-duplicatas (mesmo início ou mesmo fim): permanecem, mas formam um grupo que não pode ser separado
pai = np.arange(len(base))


def raiz(i):
    while pai[i] != i:
        pai[i] = pai[pai[i]]
        i = pai[i]
    return i


for coluna in ("chave_inicio", "chave_fim"):
    for posicoes in base.groupby(coluna).indices.values():
        for j in posicoes[1:]:
            pai[raiz(j)] = raiz(posicoes[0])
base["grupo_id"] = [raiz(i) for i in range(len(base))]

# o teste original é preservado: se um grupo toca o treino original, os registros do teste são excluídos (nunca transferidos)
tem_treino = (base["particao_original"] == "train").groupby(base["grupo_id"]).transform("max")
conflitantes = (base["particao_original"] == "test") & tem_treino
ids_treino = base[base["particao_original"] == "train"].groupby("grupo_id")["id_origem"].agg(", ".join)
excluidas = base.loc[conflitantes, ["id_origem", "grupo_id", "sentimento"]].assign(
    ids_treino_do_grupo=base.loc[conflitantes, "grupo_id"].map(ids_treino),
    motivo="quase-duplicata de avaliação do treino original")
base = base.loc[~conflitantes].reset_index(drop=True)
assert base.groupby("grupo_id")["particao_original"].nunique().max() == 1, "grupo com avaliações de treino e de teste"

print("Vínculo entre cada cópia removida e o representante mantido (amostra):")
display(removidas.head(3))
print("Exclusões do teste original por quase-duplicata de avaliação do treino (todas):")
display(excluidas.reset_index(drop=True))
pd.Series({
    "avaliações originais": len(df),
    "cópias equivalentes removidas": len(removidas),
    "  das quais removidas do teste original (representante no treino)": int(((removidas["particao_original"] == "test") & (removidas["particao_representante"] == "train")).sum()),
    "grupos descartados por sentimento conflitante": n_conflito,
    "avaliações mantidas": len(base),
    "  originárias do treino original": int((base["particao_original"] == "train").sum()),
    "  originárias do teste original": int((base["particao_original"] == "test").sum()),
    "grupos de quase-duplicatas (2 ou mais avaliações)": int((base.groupby("grupo_id").size() > 1).sum()),
    "avaliações do teste original excluídas por quase-duplicata do treino": len(excluidas),
}, name="quantidade").to_frame()
````

* **12 avaliações do teste original excluídas** por pertencerem a grupos de quase-duplicatas com avaliações do treino. A tabela de exclusões acima lista cada uma delas com os registros do treino correspondentes; nenhuma foi transferida para o treino.

### 3.2 Divisão dos dados

**Decisões e justificativas.**

* **Origem preservada:** treino e validação contêm **somente** avaliações do treino original; teste e generalização contêm **somente** avaliações do teste original. Isso é verificado por asserções na própria célula de divisão e nas validações (3.5).
* **Teste:** é a partição de teste original (após a deduplicação e a exclusão de 12 conflitos), exceto o subconjunto de generalização. Assim a separação original é respeitada, e o teste não contém textos do treino.
* **Validação:** 15% dos grupos do treino original, sorteados com semente fixa e estratificados por sentimento, para o controle de *overfitting* na Etapa de modelagem.
* **Generalização:** 1.200 grupos do teste original (600 por sentimento) são reservados para o teste complementar da seção 3.6, sem participar do treinamento nem de ajustes.
* **Proporção:** o teste ficou maior que o treino porque a partição original do IMDB é 50/50. Optou-se por não descartar dados nem reorganizar a separação original além do necessário.
* **Sem ajuste com dados de teste:** as regras de geração e seus parâmetros foram decididos a partir do desenho da solução e de diagnósticos feitos **apenas em treino e validação** (seção 3.5). Um único número da EDA entra como parâmetro de verificação, a repetição natural P99, e foi recalculado só com o treino original.

````python
grupos = base.groupby("grupo_id").agg(origem=("particao_original", "first"), sentimento=("sentimento", "first")).reset_index()
rng_divisao = np.random.default_rng(SEMENTE)
destino = {}
for (origem, sentimento), sub in grupos.groupby(["origem", "sentimento"]):
    ids = rng_divisao.permutation(sub["grupo_id"].to_numpy())
    corte = round(FRACAO_VALIDACAO * len(ids)) if origem == "train" else N_GENERALIZACAO // 2
    rotulos = ("validacao", "treino") if origem == "train" else ("generalizacao", "teste")
    destino.update(dict.fromkeys(ids[:corte], rotulos[0]))
    destino.update(dict.fromkeys(ids[corte:], rotulos[1]))
base["particao"] = base["grupo_id"].map(destino)
assert base.loc[base["particao"].isin(["treino", "validacao"]), "particao_original"].eq("train").all(), "treino/validação com avaliação do teste original"
assert base.loc[base["particao"].isin(["teste", "generalizacao"]), "particao_original"].eq("test").all(), "teste/generalização com avaliação do treino original"

ordem = ["treino", "validacao", "teste", "generalizacao"]
tabela = pd.crosstab(base["particao"], base["sentimento"]).loc[ordem]
tabela["total"] = tabela.sum(axis=1)
tabela["% do total"] = tabela["total"] / tabela["total"].sum() * 100
tabela["% positivo"] = tabela["positivo"] / tabela["total"] * 100
tabela
````

### 3.3 Geração das anomalias sintéticas

As três categorias aprovadas na Fase 1 são implementadas como **inserções de um trecho** no texto limpo da avaliação. Os trechos são sintéticos e fictícios (domínios, telefones e marcas inventados) e **não representam fraudes comprovadas**.

| Categoria | Mecanismos do treino (padrão) | Mecanismos inéditos (generalização) |
| :--- | :--- | :--- |
| **Spam promocional** | 7 modelos de mensagem combinando ganchos, ofertas, chamadas e referências externas (site, @usuário, e-mail, telefone fictício, código promocional); 4 dos 7 modelos incluem um site | relato pessoal que esconde a promoção, mensagens em outros idiomas, link ofuscado, conteúdo patrocinado |
| **Repetição excessiva** | palavra (8 a 24 vezes), expressão de 2 a 4 palavras (4 a 10 vezes) ou frase de 4 a 8 palavras (3 a 5 vezes), tiradas do próprio texto da avaliação ou de um repertório neutro | blocos alternados, alongamento de caracteres, sequência sem sentido, enumeração numerada |
| **Manipulação de avaliações** | 10 modelos de pedidos de nota, voto ou marcação de "útil" e de troca de avaliações; 70% pedem nota alta e 30% nota baixa | pedido indireto por ranking de revisores, troca mútua, revelação de pagamento, campanha coordenada, outros idiomas |

**Regras aplicadas a todas as anomalias.**

* **Seleção independente de nota e sentimento:** em cada combinação de partição e sentimento, metade das avaliações é sorteada com semente fixa; os tipos se alternam entre as sorteadas. Nem a nota, nem o sentimento, nem o comprimento participam da escolha.
* **Variedade:** os trechos são montados a partir de listas de ganchos, ofertas, chamadas e números sorteados, com variação de pontuação e de caixa. Isso evita um único modelo repetido em todas as amostras.
* **Posição variável:** o ponto de inserção é sorteado entre as fronteiras de palavra até a 160ª palavra; em metade dos casos ele é o fim de uma frase. Os trechos têm no máximo 40 palavras, de modo que a anomalia **termina até a 200ª palavra**. A Fase 2 indicou um comprimento máximo entre 256 e 400 como hipótese; a janela de 200 deixa margem para essa faixa. A tokenização definitiva ainda não existe, então a Etapa 04 deverá conferir essa garantia com as posições gravadas.
* **Manipulação nos dois sentidos:** a Fase 1 define manipulação como tentativa de inflar ou deprimir a reputação do título. Por isso uma parte dos pedidos é de nota baixa, o que impede que um "tom positivo" sirva de atalho. A maioria continua sendo de nota alta, conforme a definição do projeto.
* **Rastreabilidade:** cada anomalia registra tipo, variante, trecho inserido, posições em caracteres e em palavras. O texto original é preservado, e remover o trecho recupera exatamente o texto limpo.
* **Intensidade da repetição** acima do natural: a repetição consecutiva natural tem P99 = 3 pares (treino original), e a repetição de palavra do dataset tem ao menos 7.

````python
MARCAS = ["Zentro", "Lumavue", "Quibo", "Vantiq", "Orbello", "Kyndra", "Blixa", "Narvo"]
SLOTS_INEDITOS = {
    "marca": MARCAS, "tld": TLD,
    "pct": SLOTS_SPAM["pct"], "handle": SLOTS_SPAM["handle"],
    "indicacao": lambda rng: "".join(rng.choices("ABCDEFGHJKLMNPQRSTUVWXYZ23456789", k=6)),
    "lista": ["ranking", "hall of fame", "list of top reviewers", "leaderboard"],
    "alvo": ["the average", "its score", "the rating"],
}
MODELOS_INEDITOS = {
    "spam_promocional": {
        "camuflado": ["Funny story, I only watched this because a subscription service gave me a free month, ask me how to get yours with referral {indicacao}.",
                      "Thanks to the {marca} giveaway I saw this for free, and their sponsor page has {pct}% off for newcomers.",
                      "Watched it on a freebie from {marca}, the first month costs nothing if you mention my name."],
        "multilingue": ["Ganhe {pct}% de desconto assinando hoje, link na bio!", "Películas gratis en HD, solo hoy, escribe a {handle}.",
                        "Jetzt kostenlos streamen und {pct}% sparen bei {marca}!", "Oferta exclusiva da {marca}: filmes sem pagar nada, chame no WhatsApp."],
        "link_ofuscado": ["{marca} [dot] com has every new release in HD, no signup needed.", "w w w dot {marca} dot c o m for cheap tickets.",
                          "h t t p : / / {marca} . {tld} (copy the address into your browser)."],
        "patrocinado": ["Sponsored content: this review is brought to you by {marca}, our official partner.",
                        "Paid partnership with {marca}, details on their official page.", "Ad: {marca} supports this channel, explore their catalog."],
    },
    "manipulacao_avaliacoes": {
        "ranking_pessoal": ["Marking this helpful keeps me on the reviewer {lista}, thank you kindly.",
                            "Every helpful vote moves my profile up the {lista}, so thanks in advance."],
        "troca_mutua": ["Follow my profile and I will follow back and score every one of your reviews.", "Swap deal: you boost my reviews, I boost yours."],
        "pago": ["Full disclosure, the studio asked me to push this one to a top rating, so consider doing the same.",
                 "I was told to bring {alvo} up before the weekend, join in."],
        "campanha": ["Coordinated review-bombing starts tonight, everyone drop their scores to the bottom.",
                     "The one-star campaign is on, spread the word and sink {alvo}."],
        "multilingue": ["Por favor, deem nota 10 para este filme!", "Vota una estrella para hundirla, entre todos podemos.", "Bitte gebt zehn Punkte, das hilft uns sehr."],
    },
}


def gerar_inedito(tipo):
    def gerar(palavras, rng):
        variante = rng.choice(list(MODELOS_INEDITOS[tipo]))
        return variante, preencher(rng.choice(MODELOS_INEDITOS[tipo][variante]), SLOTS_INEDITOS, rng)
    return gerar


def gerar_repeticao_inedita(palavras, rng):
    variante = rng.choice(["blocos", "alongamento", "sem_sentido", "enumeracao"])
    if variante == "blocos":      # A A A B B B A A A B B B
        a, b = rng.sample(["yes", "no", "up", "down", "left", "right", "fast", "slow", "on", "off"], 2)
        k = rng.randint(2, 3)
        return variante, " ".join(([a] * k + [b] * k) * rng.randint(2, 3))
    if variante == "alongamento":  # repetição de caracteres dentro das palavras
        alvos = rng.sample(["so", "love", "good", "wow", "please", "yes", "best", "more"], rng.randint(3, 5))
        return variante, " ".join(w[:1] + w[1] * rng.randint(6, 14) + w[2:] for w in alvos)
    if variante == "sem_sentido":  # sequência sem significado repetida
        return variante, " ".join(["".join(rng.choices("abcdefghijklmnopqrstuvwxyz", k=rng.randint(3, 6)))] * rng.randint(6, 15))
    objeto = rng.choice(["film", "movie", "scene", "actor", "part"])
    return variante, " ".join(f"{i} {objeto}" for i in range(1, rng.randint(6, 12) + 1))


GERADORES_INEDITOS = {"spam_promocional": gerar_inedito("spam_promocional"), "repeticao_excessiva": gerar_repeticao_inedita,
                      "manipulacao_avaliacoes": gerar_inedito("manipulacao_avaliacoes")}
````

````python
def estilizar(trecho, rng):
    """Varia pontuação (metade dos '!' viram '.') e caixa: MAIÚSCULAS (6%), minúsculas (25%) ou como escrito."""
    if rng.random() < 0.5:
        trecho = trecho.replace("!", ".")
    r = rng.random()
    return trecho.upper() if r < 0.06 else trecho.lower() if r < 0.31 else trecho


def escolher_posicao(palavras, rng):
    """Posição de inserção (em palavras): metade das vezes em fim de frase, o restante em qualquer fronteira."""
    limite = min(len(palavras), POS_MAX)
    if rng.random() < 0.5:
        fins = [i + 1 for i, w in enumerate(palavras[:limite]) if w.endswith((".", "!", "?"))]
        if fins:
            return rng.choice(fins)
    return rng.randint(0, limite)


def injetar(palavras, gerador, rng):
    """Devolve (texto, trecho, início, fim, posição em palavras, variante)."""
    for _ in range(10):
        variante, trecho = gerador(palavras, rng)
        segmento = estilizar(trecho, rng).split()
        if 0 < len(segmento) <= SEG_MAX:
            break
    else:
        raise ValueError("trecho fora do tamanho permitido")
    pos = escolher_posicao(palavras, rng)
    antes = " ".join(palavras[:pos])
    ini = len(antes) + (1 if pos > 0 else 0)
    trecho = " ".join(segmento)
    return " ".join(palavras[:pos] + segmento + palavras[pos:]), trecho, ini, ini + len(trecho), pos, variante


def atribuir_classes(base, semente):
    """Metade de cada (partição, sentimento) recebe anomalia; os tipos se alternam entre as escolhidas."""
    ds = base.sort_values("id_origem").reset_index(drop=True)
    ds["classe"], ds["tipo_anomalia"] = 0, "nenhuma"
    rng = np.random.default_rng([semente, 1])
    indices = ds.groupby(["particao", "sentimento"]).indices
    for chave in sorted(indices):
        escolhidos = rng.permutation(indices[chave])[: len(indices[chave]) // 2]
        ds.loc[escolhidos, "classe"] = 1
        ds.loc[escolhidos, "tipo_anomalia"] = np.resize(TIPOS, len(escolhidos))
    return ds


def gerar_textos(ds, semente):
    saidas = []
    for id_origem, texto, tipo, particao in zip(ds["id_origem"], ds["texto_original"], ds["tipo_anomalia"], ds["particao"]):
        palavras = limpar(texto).split(" ")
        if tipo == "nenhuma":
            saidas.append((" ".join(palavras), "", -1, -1, -1, ""))
        else:
            geradores = GERADORES_INEDITOS if particao == "generalizacao" else GERADORES_PADRAO
            saidas.append(injetar(palavras, geradores[tipo], random.Random(f"{semente}|{id_origem}")))
    return pd.DataFrame(saidas, columns=["texto_modelo", "trecho_inserido", "ini_anomalia", "fim_anomalia", "pos_palavra", "variante"], index=ds.index)


COLUNAS = ["id_origem", "grupo_id", "n_copias_removidas", "particao_original", "particao", "sentimento", "texto_original",
           "texto_modelo", "classe", "tipo_anomalia", "variante", "sintetico", "trecho_inserido", "ini_anomalia", "fim_anomalia", "pos_palavra"]


def construir_dataset(base, semente):
    ds = atribuir_classes(base, semente)
    ds = ds.join(gerar_textos(ds, semente))
    ds["sintetico"] = ds["classe"] == 1
    return ds[COLUNAS]


dataset = construir_dataset(base, SEMENTE)
dataset.shape
````

Amostras reproduzíveis (semente 42) das três categorias, com inserções no início (palavras 0 a 20), no meio (21 a 100) e no final da janela (101 a 160). No texto original, ‸ marca o ponto de inserção; no texto modificado, o trecho inserido aparece entre « ». O original traz `<br />`, que é normalizado no texto do classificador. Em seguida há uma amostra de cada categoria com mecanismos inéditos (generalização) e a proporção de inserções que ocorrem no início do texto ou logo após o fim de uma frase.

````python
def janela(linha, margem=110):
    """Contexto antes e depois da inserção; ‸ marca o ponto de inserção no texto original limpo."""
    t, ini, fim = linha["texto_modelo"], linha["ini_anomalia"], linha["fim_anomalia"]
    original = limpar(linha["texto_original"])
    cortes = ("..." if ini > margem else "", "..." if fim + margem < len(t) else "")
    antes = cortes[0] + original[max(0, ini - margem):ini] + "‸" + original[ini:ini + margem] + cortes[1]
    depois = cortes[0] + t[max(0, ini - margem):ini] + "«" + t[ini:fim] + "»" + t[fim:fim + margem] + cortes[1]
    return antes, depois


faixas = {"início (palavras 0 a 20)": (0, 20), "meio (21 a 100)": (21, 100), "final da janela (101 a 160)": (101, 160)}
anomalias_treino = dataset[(dataset["classe"] == 1) & (dataset["particao"] == "treino")]
for tipo in TIPOS:
    for nome, (a, b) in faixas.items():
        linha = anomalias_treino[(anomalias_treino["tipo_anomalia"] == tipo) & anomalias_treino["pos_palavra"].between(a, b)].sample(1, random_state=SEMENTE).iloc[0]
        antes, depois = janela(linha)
        print(f"[{tipo} | {nome} | variante '{linha['variante']}' | {linha['sentimento']} | posição {linha['pos_palavra']}]")
        print("  ANTES :", antes)
        print("  DEPOIS:", depois, "\n")

print("Mecanismos inéditos (generalização), uma amostra por categoria:")
for tipo in TIPOS:
    linha = dataset[(dataset["particao"] == "generalizacao") & (dataset["tipo_anomalia"] == tipo)].sample(1, random_state=SEMENTE).iloc[0]
    print(f"[{tipo} | variante '{linha['variante']}' | {linha['sentimento']} | posição {linha['pos_palavra']}]")
    print("  DEPOIS:", janela(linha)[1], "\n")

apos_frase = anomalias_treino.apply(lambda l: l["ini_anomalia"] == 0 or l["texto_modelo"][:l["ini_anomalia"]].rstrip().endswith((".", "!", "?")), axis=1)
print("Inserções no início do texto ou logo após o fim de uma frase (%), por tipo:")
(apos_frase.groupby(anomalias_treino["tipo_anomalia"]).mean() * 100).to_frame("%")
````

* **Consistência:** em todas as amostras o trecho aparece íntegro, na posição indicada, e o restante do texto é idêntico ao original limpo. A reconstrução exata do texto original ao remover o trecho também é verificada automaticamente em 100% das anomalias (3.5).
* **Posição:** entre 52% e 54% das inserções ocorrem no início do texto ou logo após o fim de uma frase (metade por desenho, mais os casos aleatórios que coincidem com fim de frase); as demais caem no meio de frases (por exemplo, "the worst acting I've «...» ever seen"), como ocorreria numa injeção real.
* **Problemas observados, não corrigidos:** (1) algumas combinações de spam são agramaticais, como "subscribe to our channel for buy one get one free", porque o modelo de mensagem e a oferta são sorteados separadamente; (2) a "frase" repetida é uma janela de palavras consecutivas que pode começar no meio de uma frase ("by klaus kinski as reverend cotton is ..."), então o nome é aproximado; (3) a repetição de expressão tirada do próprio texto pode formar sequências sem sentido ("fairy godparents because basically ..."), o que é esperado em texto artificial; (4) a variação de caixa e pontuação altera a aparência de parte dos trechos (25% em minúsculas, 6% em maiúsculas).
````python
DETECTORES = {
    "spam_promocional": re.compile(r"www\.|https?:|h\s*t\s*t\s*p|\.(?:com|net|org|biz|info|tv|co|xyz)\b|@\w|%|free|discount|coupon|promo|code|offer|deal|subscribe|buy|click|visit|download|win\b|bio\b|call|email|whatsapp|price|membership|order|sign up|sponsor|referral|giveaway|\bdot\b|gratis|grátis|desconto|descuento|sparen|oferta|\bad:|partnership", re.I),
    "manipulacao_avaliacoes": re.compile(r"vot|rat(?:e|ing)|stars?|\d\s*/\s*10|helpful|score|upvote|downvote|thumbs|leaderboard|follow|nota|estrella|punkte|review-bombing|top rating|average|marks|out of ten|boost|swap|give (?:this|it)|a 10", re.I),
}


def tem_repeticao(trecho):
    tokens = re.findall(r"\w+", trecho.lower())
    diversidade = len(set(tokens)) / max(len(tokens), 1)
    return diversidade <= 0.5 or max(Counter(tokens).values(), default=0) >= 5 or bool(re.search(r"(\w)\1{4,}", trecho.lower()))


def padrao_presente(linha):
    if linha["tipo_anomalia"] == "repeticao_excessiva":
        return tem_repeticao(linha["trecho_inserido"])
    return bool(DETECTORES[linha["tipo_anomalia"]].search(linha["trecho_inserido"]))


def remover_trecho(linha):
    t, ini, fim = linha["texto_modelo"], linha["ini_anomalia"], linha["fim_anomalia"]
    return t[:ini] + t[fim + 1:] if fim < len(t) else t[:ini - 1]


sint = dataset[dataset["classe"] == 1]
normal = dataset[dataset["classe"] == 0]
ids_df = df["particao"] + "/" + df["sentimento"].str[:3] + "/" + df["id_review"].astype(str) + "_" + df["nota"].astype(str)
originais = pd.Series(df["texto"].to_numpy(), index=ids_df.to_numpy())
p99_natural = df.loc[df["particao"] == "train", "rep_consec"].quantile(0.99)   # calculado só com o treino original
reps_palavra = sint.query("tipo_anomalia == 'repeticao_excessiva' and variante == 'palavra'")["trecho_inserido"].map(lambda t: len(re.findall(r"\w+", t)))
cont = dataset.groupby(["particao", "sentimento", "classe"]).size().unstack()
tipos_cont = sint.groupby(["particao", "sentimento", "tipo_anomalia"]).size().unstack()
pos_pct = dataset.groupby(["particao", "classe"])["sentimento"].apply(lambda s: (s == "positivo").mean())
reconstruidos = sint.apply(remover_trecho, axis=1)
palavras_ate_fim = sint.apply(lambda l: len(l["texto_modelo"][:l["fim_anomalia"]].split()), axis=1)
dataset_2 = construir_dataset(base, SEMENTE)
dataset_3 = construir_dataset(base, SEMENTE + 1)
hash_ds = lambda d: pd.util.hash_pandas_object(d[["id_origem", "texto_modelo", "classe", "tipo_anomalia"]], index=False).sum()
fracao_unica = sint.groupby("tipo_anomalia")["trecho_inserido"].agg(lambda s: s.value_counts().iloc[0] / len(s))
chi_p = [chi2_contingency(pd.crosstab(d["classe"], d["sentimento"]))[1] for _, d in dataset.groupby("particao")]
em_treino_val = dataset["particao"].isin(["treino", "validacao"])

ger = dataset["particao"] == "generalizacao"
inseridos = dataset["classe"] == 1
ids_validos = dataset["id_origem"].str.fullmatch(r"(train|test)/(pos|neg)/\d+_\d+")
ids_treino_conflito = excluidas["ids_treino_do_grupo"].str.split(", ").explode()
sem_bordas = dataset["texto_modelo"].map(lambda t: t == t.strip() and "  " not in t and "<br" not in t).all()

verificacoes = [
    ("Identificadores válidos, únicos, presentes no IMDB carregado e coerentes com sentimento e partição original",
     bool(dataset["id_origem"].is_unique and ids_validos.all() and dataset["id_origem"].isin(originais.index).all()
          and (dataset["id_origem"].str.split("/").str[1] == dataset["sentimento"].str[:3]).all()
          and (dataset["id_origem"].str.split("/").str[0] == dataset["particao_original"]).all()), f"{len(dataset):,} linhas"),
    ("Nenhum texto equivalente (chave normalizada) se repete", dataset["texto_original"].map(chave_normalizada).is_unique, "duplicatas removidas"),
    ("Nenhum grupo de quase-duplicatas atravessa partições", dataset.groupby("grupo_id")["particao"].nunique().max() == 1, "agrupamento por início/fim"),
    ("Treino e validação só do treino original; teste e generalização só do teste original",
     bool(dataset.loc[em_treino_val, "particao_original"].eq("train").all() and dataset.loc[~em_treino_val, "particao_original"].eq("test").all()), "sem transferência entre origens"),
    ("Exclusões rastreáveis: ausentes do dataset e ligadas a registros do treino presentes",
     bool(not excluidas["id_origem"].isin(dataset["id_origem"]).any() and ids_treino_conflito.isin(dataset.loc[em_treino_val, "id_origem"]).all()), f"{len(excluidas)} exclusões"),
    ("Sem textos vazios; texto do classificador sem <br, espaços duplicados ou nas bordas",
     bool((dataset["texto_original"].str.strip() != "").all() and (dataset["texto_modelo"].str.strip() != "").all() and sem_bordas), "0 vazios"),
    ("Classes equilibradas por partição e sentimento (diferença <= 1)", bool((cont[0] - cont[1]).abs().max() <= 1), f"diferença máx. = {(cont[0] - cont[1]).abs().max()}"),
    ("Tipos equilibrados por partição e sentimento (diferença <= 1)", bool((tipos_cont.max(axis=1) - tipos_cont.min(axis=1)).max() <= 1), f"diferença máx. = {(tipos_cont.max(axis=1) - tipos_cont.min(axis=1)).max()}"),
    ("% de positivos igual nas duas classes (diferença < 1 p.p.)", bool((pos_pct.unstack().diff(axis=1)[1].abs() < 0.01).all()), f"p-valores qui-quadrado: {[round(float(p), 2) for p in chi_p]}"),
    ("Rótulos coerentes: classe, sintetico, tipo e trecho inserido",
     bool(((dataset["classe"] == 1) == dataset["sintetico"]).all() and (inseridos == (dataset["tipo_anomalia"] != "nenhuma")).all() and (inseridos == (dataset["trecho_inserido"] != "")).all()), "nenhum sintético sem trecho"),
    ("Padrão inserido existe em 100% das anomalias", bool(sint.apply(padrao_presente, axis=1).all()), f"{len(sint):,} anomalias"),
    ("Trecho registrado coincide com as posições gravadas", bool(sint.apply(lambda l: l["texto_modelo"][l["ini_anomalia"]:l["fim_anomalia"]] == l["trecho_inserido"], axis=1).all()), "início/fim corretos"),
    ("Remover o trecho recupera o texto limpo original", bool((reconstruidos == sint["texto_original"].map(limpar)).all()), "reconstrução exata"),
    ("Classe 0: texto do classificador = texto original limpo", bool((normal["texto_modelo"] == normal["texto_original"].map(limpar)).all()), "sem alteração indevida"),
    ("Textos originais idênticos aos carregados na Fase 2", bool((originais.loc[dataset["id_origem"]].to_numpy() == dataset["texto_original"].to_numpy()).all()) and pd.util.hash_pandas_object(df["texto"], index=False).sum() == HASH_TEXTOS, "hash do df conferido"),
    ("Anomalia termina dentro da janela segura", bool((palavras_ate_fim <= JANELA_SEGURA).all()), f"máx. = {palavras_ate_fim.max()} palavras (janela {JANELA_SEGURA})"),
    ("Repetição de palavra supera a repetição natural (P99 do treino)", bool((reps_palavra - 1 > p99_natural).all()), f"mín. {reps_palavra.min() - 1} pares consecutivos > P99 = {p99_natural:.0f}"),
    ("Nenhum trecho domina o seu tipo (< 2%)", bool((fracao_unica < 0.02).all()), f"maior repetição: {fracao_unica.max():.2%}"),
    ("Mecanismos e trechos da generalização ausentes do treino, validação e teste",
     bool(set(dataset.loc[ger & inseridos, "variante"]).isdisjoint(dataset.loc[~ger & inseridos, "variante"]) and set(dataset.loc[ger & inseridos, "trecho_inserido"]).isdisjoint(dataset.loc[~ger & inseridos, "trecho_inserido"])), "variantes e trechos disjuntos"),
    ("Reprodutível com a mesma semente", hash_ds(dataset) == hash_ds(dataset_2), "hashes idênticos"),
    ("Semente diferente altera as anomalias", bool((dataset_3.loc[dataset_3["classe"] == 1, "texto_modelo"].to_numpy() != dataset.loc[dataset["classe"] == 1, "texto_modelo"].to_numpy()).mean() > 0.9), "> 90% dos textos mudam"),
]
resultado = pd.DataFrame(verificacoes, columns=["verificação", "ok", "detalhe"])
display(resultado)
assert resultado["ok"].all(), resultado[~resultado["ok"]]
````

````python
URL_RX = r"www\.|https?:|\b[\w-]+\.(?:com|net|org|biz|info|tv|co|xyz)\b"
POSITIVAS = {"great", "good", "best", "excellent", "amazing", "love", "loved", "wonderful", "perfect", "brilliant", "superb", "fantastic"}
NEGATIVAS = {"bad", "worst", "awful", "terrible", "boring", "waste", "poor", "stupid", "horrible", "dull", "disappointing", "mediocre"}


def polaridade(texto):
    tokens = tokenizar(texto)
    return 100 * (sum(t in POSITIVAS for t in tokens) - sum(t in NEGATIVAS for t in tokens)) / max(len(tokens), 1)


dev = dataset[em_treino_val].copy()    # diagnóstico apenas com treino e validação
assert set(dev["particao"]) == {"treino", "validacao"}, "o diagnóstico não pode usar teste nem generalização"
t = dev["texto_modelo"]
carac = pd.DataFrame({
    "palavras (texto do classificador)": t.str.count(" ") + 1,
    "palavras (texto original)": dev["texto_original"].str.split().str.len(),
    "caracteres": t.str.len(),
    "contém URL": t.str.contains(URL_RX, case=False),
    "dígitos": t.str.count(r"\d"),
    "exclamações": t.str.count("!"),
    "fração de letras maiúsculas": t.str.count(r"[A-Z]") / t.str.count(r"[A-Za-z]"),
    "palavra em MAIÚSCULAS (4+ letras)": t.str.contains(r"\b[A-Z]{4,}\b"),
    "caractere não ASCII": t.str.contains(r"[^\x00-\x7f]"),
    "começa com maiúscula": t.str.match(r"[A-Z]"),
    "termina com . ! ?": t.str.contains(r"[.!?]$"),
    "polaridade lexical": t.map(polaridade),
    "<br /> remanescente": t.str.contains("<br", regex=False),
}).astype(float)


def auc(coluna, mascara=None):
    m = slice(None) if mascara is None else mascara
    return roc_auc_score(dev["classe"][m], carac[coluna][m])


linhas = {c: {"média (classe 0)": carac.loc[dev["classe"] == 0, c].mean(), "média (classe 1)": carac.loc[dev["classe"] == 1, c].mean(),
              "AUC": auc(c), "AUC só positivos": auc(c, dev["sentimento"] == "positivo"), "AUC só negativos": auc(c, dev["sentimento"] == "negativo")}
          for c in carac.columns}
atalhos = pd.DataFrame(linhas).T
atalhos["|AUC - 0,5| >= 0,10"] = (atalhos["AUC"] - 0.5).abs() >= 0.10
display(atalhos)

print("Tipo de anomalia por sentimento (% de positivos em cada tipo):")
display((dev[dev["classe"] == 1].groupby("tipo_anomalia")["sentimento"].apply(lambda s: (s == "positivo").mean() * 100)).to_frame("% positivos"))
````

### 4.2 Padronização e tokenização

**Ferramenta.** Foi usada a camada `TextVectorization` do Keras, por quatro motivos ligados à explicabilidade futura: recebe **texto bruto**, o que permite chamá-la a partir do `classifier_fn` do LIME e da interface Gradio; a padronização é feita com operações do TensorFlow, de modo que pode ser embutida no modelo; o vocabulário é ajustado com `adapt` somente nos dados fornecidos; e o truncamento mantém o **início** do texto, com *padding* à direita, compatível com a máscara da LSTM.

**Padronização** (função `padronizar`):

* converte para minúsculas e remove `<br />`;
* normaliza aspas e apóstrofos tipográficos;
* **mantém contrações** como um único token (*don't*, *isn't*), além de acentos e dígitos;
* transforma `! ? % $ @ /` em **tokens próprios**, porque são sinais de spam e de manipulação (promoções, contatos, notas "10/10");
* converte as demais pontuações em espaço, de modo que um endereço vira partes como `www`, nome e `com`.

**Stopwords não são removidas.** As negações (*not*, *no*, *never*, *nor*, *cannot*) e as palavras funcionais fazem parte dos sinais que o modelo precisa ver, e a repetição excessiva frequentemente envolve palavras funcionais. A remoção de stopwords é citada no enunciado como exemplo de limpeza, mas aqui é uma decisão consciente do grupo.

**Palavras desconhecidas** recebem o índice 1 (`[UNK]`), e o *padding* usa o índice 0.

As duas células a seguir ajustam um vocabulário completo apenas com o treino, mostram um exemplo de tokenização e verificam a presença dos sinais relevantes.

````python
@keras.saving.register_keras_serializable(package="cp2")
def padronizar(textos):
    """Minúsculas; !?%$@/ viram tokens próprios; apóstrofos internos e acentos são mantidos; demais pontuações viram espaço."""
    t = tf.strings.lower(textos, encoding="utf-8")
    t = tf.strings.regex_replace(t, r"<br\s*/?>", " ")
    t = tf.strings.regex_replace(t, "[’‘`´]", "'")
    t = tf.strings.regex_replace(t, r"([!?%$@/])", r" \1 ")
    t = tf.strings.regex_replace(t, r"[^\pL\pN\s'_!?%$@/]", " ")
    return tf.strings.regex_replace(t, r"(^|\s)'+|'+(\s|$)", " ")


def criar_camada(max_tokens=None, comprimento=None):
    return keras.layers.TextVectorization(max_tokens=max_tokens, standardize=padronizar, split="whitespace",
                                          output_mode="int", output_sequence_length=comprimento)


def em_lotes(textos, tamanho=1024):
    return tf.data.Dataset.from_tensor_slices(np.asarray(list(textos), dtype=object)).batch(tamanho)


P = {p: dataset[dataset["particao"] == p] for p in ordem}
completa = criar_camada()                                  # vocabulário completo, ajustado só com o treino
completa.adapt(em_lotes(P["treino"]["texto_modelo"]))
vocab_completo = [str(p) for p in completa.get_vocabulary()]


def ids_completos(textos, lote=512):
    """Sequências completas (sem truncar) como lista de arrays; 0 = padding, 1 = desconhecido, 2+ = palavras do treino."""
    saida = []
    for i in range(0, len(textos), lote):
        bloco = completa(tf.constant(textos[i:i + lote])).numpy()
        saida += [linha[:np.count_nonzero(linha)] for linha in bloco]
    return saida


ids_dev = {p: ids_completos(list(P[p]["texto_modelo"])) for p in ("treino", "validacao")}   # só treino e validação nesta fase
print(f"Vocabulário completo do treino: {len(vocab_completo) - 2:,} palavras distintas (fora [PAD] e [UNK])")
exemplo = P["treino"]["texto_modelo"].iloc[0]
print("\nTexto:", exemplo[:150], "...")
print("Tokens:", [vocab_completo[i] for i in ids_dev["treino"][0][:30]])
````

### 4.3 Avaliação dos vocabulários e comprimentos

Para cada combinação de vocabulário (10.000 e 20.000) e comprimento (256 e 400) são medidos, **somente com treino e validação**:

* a taxa de tokens desconhecidos, no texto inteiro e dentro dos trechos inseridos;
* o impacto do truncamento: textos truncados, tokens descartados e posições de *padding*;
* a preservação dos trechos inseridos: íntegros, cortados parcialmente ou perdidos. A posição de cada trecho em tokens é medida pelo próprio tokenizador.

Como o vocabulário é ordenado por frequência, cada vocabulário candidato é obtido a partir do completo ajustado no treino (os demais índices viram `[UNK]`); essa equivalência é verificada em 4.5. Em seguida, as regras da abertura são aplicadas.

````python
def limites_tokens(sub):
    """Posição (em tokens) em que cada trecho inserido começa e termina, medida pelo próprio tokenizador."""
    a = sub[sub["classe"] == 1]
    antes = [len(x) for x in ids_completos([t[:i] for t, i in zip(a["texto_modelo"], a["ini_anomalia"])])]
    ate = [len(x) for x in ids_completos([t[:f] for t, f in zip(a["texto_modelo"], a["fim_anomalia"])])]
    return np.array(antes), np.array(ate)


def trechos_ids(ids_lista, sub, antes, ate):
    idx = np.flatnonzero(sub["classe"].to_numpy() == 1)
    return [ids_lista[i][a:b] for i, a, b in zip(idx, antes, ate)]


lim = {p: limites_tokens(P[p]) for p in ("treino", "validacao")}
segmentos = {p: trechos_ids(ids_dev[p], P[p], *lim[p]) for p in ("treino", "validacao")}
tokens_por_texto = {p: np.array([len(x) for x in ids_dev[p]]) for p in ids_dev}
pd.DataFrame({p: pd.Series(tokens_por_texto[p]).describe(percentiles=[0.5, 0.9, 0.95, 0.99]) for p in tokens_por_texto})
````

````python
def taxa_unk(ids_lista, V):
    todos = np.concatenate(ids_lista)
    return 100 * np.mean((todos == 1) | (todos >= V))


def unk_trechos(segs, V):
    todos = np.concatenate(segs)
    return 100 * np.mean((todos == 1) | (todos >= V))


def truncamento(n_tokens, L):
    return {"textos truncados (%)": 100 * np.mean(n_tokens > L), "tokens descartados (%)": 100 * np.maximum(n_tokens - L, 0).sum() / n_tokens.sum(),
            "posições de padding (%)": 100 * (1 - np.minimum(n_tokens, L).sum() / (len(n_tokens) * L))}


def preservacao(antes, ate, L):
    return {"trechos íntegros (%)": 100 * np.mean(ate <= L), "trechos parciais (%)": 100 * np.mean((antes < L) & (ate > L)), "trechos perdidos (%)": 100 * np.mean(antes >= L)}


antes_dev, ate_dev = np.concatenate([lim["treino"][0], lim["validacao"][0]]), np.concatenate([lim["treino"][1], lim["validacao"][1]])
linhas = []
for V in VOCABULARIOS:
    for L in COMPRIMENTOS:
        linhas.append({"vocabulário": V, "comprimento": L,
                       "UNK treino (%)": taxa_unk(ids_dev["treino"], V), "UNK validação (%)": taxa_unk(ids_dev["validacao"], V),
                       "UNK nos trechos, validação (%)": unk_trechos(segmentos["validacao"], V),
                       **truncamento(tokens_por_texto["validacao"], L), **preservacao(antes_dev, ate_dev, L)})
avaliacao = pd.DataFrame(linhas)
display(avaliacao.set_index(["vocabulário", "comprimento"]).T)

# regras decididas antes de olhar os resultados, aplicadas apenas a treino e validação
unk_ref = unk_trechos(segmentos["validacao"], max(VOCABULARIOS))
V_ESCOLHIDO = min(V for V in VOCABULARIOS if unk_trechos(segmentos["validacao"], V) - unk_ref <= 100 * MAX_AUMENTO_UNK)
L_ESCOLHIDO = min((L for L in COMPRIMENTOS if preservacao(antes_dev, ate_dev, L)["trechos perdidos (%)"] + preservacao(antes_dev, ate_dev, L)["trechos parciais (%)"] <= 100 * MAX_PERDA_ANOMALIA), default=max(COMPRIMENTOS))
print(f"Configuração escolhida pelas regras: vocabulário = {V_ESCOLHIDO:,} | comprimento = {L_ESCOLHIDO}")
````

### 4.6 Compatibilidade com LIME e Gradio

O LIME chama o classificador com uma **lista de textos brutos**, formada por versões do texto original com palavras removidas, e espera uma probabilidade por texto. O Gradio entrega um único texto bruto. A célula abaixo verifica que `vetorizar`:

* aceita um texto ou uma lista de textos e devolve sempre `(n, 256)`;
* dá o mesmo resultado em lote e individualmente (condição para o LIME, que envia milhares de variações em lote);
* lida com perturbações (palavras removidas), texto vazio, só pontuação, texto longo e acentos e símbolos.

````python
linha = P["validacao"][P["validacao"]["classe"] == 1].iloc[0]
texto = linha["texto_modelo"]
ids = vetorizar(texto)[0]
antes, ate = limites_tokens(P["validacao"].loc[[linha.name]])
print(f"Texto ({linha['tipo_anomalia']}): ...{texto[max(0, linha['ini_anomalia'] - 60):linha['fim_anomalia'] + 40]}...")
print("Tokens do trecho inserido:", decodificar(ids[antes[0]:ate[0]]))

rng_demo = random.Random(SEMENTE)
palavras = texto.split()
perturbados = [" ".join(w for w in palavras if rng_demo.random() > 0.3) for _ in range(4)]    # perturbação semelhante à do LIME
casos = {"texto original": texto, "perturbação 1 (palavras removidas)": perturbados[0], "perturbação 2": perturbados[1], "texto vazio": "",
         "só pontuação": "!!! ... ???", "texto muito longo": " ".join([texto] * 4), "acentos e símbolos": "Ganhe 50% de desconto! www.exemplo.com"}
matriz = vetorizar(list(casos.values()))
isolado = np.stack([vetorizar(t)[0] for t in casos.values()])
print("\nSaída em lote igual à saída individual (independência entre textos):", bool((matriz == isolado).all()))
pd.DataFrame({"tokens não nulos": (matriz != 0).sum(axis=1), "UNK": (matriz == 1).sum(axis=1), "formato da linha": [matriz.shape[1]] * len(casos)}, index=list(casos))
````

### 5.1 Baselines

Os baselines são treinados com o treino e avaliados na validação.

````python
import time
from math import ceil

from scipy import sparse
from sklearn.dummy import DummyClassifier
from sklearn.feature_extraction.text import TfidfTransformer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, f1_score, log_loss, precision_score, recall_score, roc_auc_score

DIM_EMBEDDING = 64
UNIDADES_LSTM = 64
TAXA_DROPOUT = 0.5
TAXA_APRENDIZADO = 1e-3
TAMANHO_LOTE = 64
MAX_EPOCAS = 15
PACIENCIA = 3

dados = {p: (X[p], y[p]) for p in ("treino", "validacao")}    # nesta etapa só estas duas partições são acessadas
(X_tr, y_tr), (X_va, y_va) = dados["treino"], dados["validacao"]
tipos_va = P["validacao"]["tipo_anomalia"].to_numpy()
variantes_va = P["validacao"]["variante"].to_numpy()


def metricas(y_real, prob):
    previsto = (prob >= 0.5).astype(int)
    return {"accuracy": accuracy_score(y_real, previsto), "precision": precision_score(y_real, previsto, zero_division=0),
            "recall": recall_score(y_real, previsto), "F1": f1_score(y_real, previsto), "ROC AUC": roc_auc_score(y_real, prob),
            "log loss": log_loss(y_real, np.clip(prob, 1e-7, 1 - 1e-7))}


print(f"Treino: X {X_tr.shape}, y {y_tr.shape} | Validação: X {X_va.shape}, y {y_va.shape} | vocabulário {V_ESCOLHIDO:,}, comprimento {L_ESCOLHIDO}")
````

````python
def bolsa_de_palavras(ids):
    """Contagem de cada índice do vocabulário por texto (sem o índice 0, que é padding)."""
    linhas = np.repeat(np.arange(len(ids)), ids.shape[1])
    contagens = sparse.csr_matrix((np.ones(ids.size, dtype="float32"), (linhas, ids.ravel())), shape=(len(ids), V_ESCOLHIDO))
    return contagens[:, 1:]


tfidf = TfidfTransformer(sublinear_tf=True)                   # ajustado só com o treino
B_tr = tfidf.fit_transform(bolsa_de_palavras(X_tr))
B_va = tfidf.transform(bolsa_de_palavras(X_va))
ingenuo = DummyClassifier(strategy="prior").fit(B_tr, y_tr)
regressao = LogisticRegression(C=1.0, solver="liblinear", max_iter=1000, random_state=SEMENTE).fit(B_tr, y_tr)

p_ing_va = ingenuo.predict_proba(B_va)[:, 1]
p_reg_tr, p_reg_va = regressao.predict_proba(B_tr)[:, 1], regressao.predict_proba(B_va)[:, 1]
pd.DataFrame({"ingênuo (validação)": metricas(y_va, p_ing_va), "regressão logística (treino)": metricas(y_tr, p_reg_tr),
              "regressão logística (validação)": metricas(y_va, p_reg_va)})
````

**Interpretação.** O ingênuo acerta 50% (a validação é balanceada) e não detecta nenhuma anomalia. A regressão logística chega a **80% de acurácia** na validação (F1 de 0,78, ROC AUC de 0,86), mas com uma diferença visível entre treino e validação (acurácia de 0,87 e AUC de 0,94 no treino), sinal de ajuste excessivo às palavras do treino. Esse é o patamar que uma LSTM precisa superar para justificar a ordem das palavras.

### 5.2 Arquitetura e hiperparâmetros

| Camada | Configuração | Parâmetros |
| :--- | :--- | ---: |
| Embedding | 20.000 tokens x 64 dimensões, `mask_zero=True` (o índice 0, de *padding*, é mascarado) | 1.280.000 |
| LSTM | 64 unidades, devolve apenas o último estado | 33.024 |
| Dropout | taxa de 0,5 | 0 |
| Dense | 1 unidade, ativação sigmoid (probabilidade da classe 1) | 65 |
| **Total** | | **1.313.089** |

Hiperparâmetros: perda *Binary Crossentropy*; otimizador Adam; lote de 64; no máximo 15 épocas; *early stopping* na perda de validação com paciência de 3 e restauração dos pesos da melhor época; limiar de decisão de 0,5. A taxa de aprendizado é definida na seção 5.3. A arquitetura é pequena de propósito (a maior parte dos parâmetros está no *embedding*) e não houve busca ampla de hiperparâmetros.

Como `padding` é feito à direita com o índice 0 e o *embedding* mascara esse índice, a LSTM ignora as posições de preenchimento. Isso é conferido em 5.7. Também é verificada a reprodutibilidade, treinando duas vezes, por uma época, com a mesma semente.

````python
def construir_modelo(taxa=TAXA_APRENDIZADO, clipnorm=None):
    keras.utils.set_random_seed(SEMENTE)
    modelo = keras.Sequential([
        keras.Input(shape=(L_ESCOLHIDO,), dtype="int32"),
        keras.layers.Embedding(V_ESCOLHIDO, DIM_EMBEDDING, mask_zero=True),     # índice 0 (padding) é mascarado
        keras.layers.LSTM(UNIDADES_LSTM),
        keras.layers.Dropout(TAXA_DROPOUT),
        keras.layers.Dense(1, activation="sigmoid"),
    ])
    modelo.compile(optimizer=keras.optimizers.Adam(taxa, clipnorm=clipnorm), loss=keras.losses.BinaryCrossentropy(),
                   metrics=["accuracy", keras.metrics.AUC(name="auc")])
    return modelo


modelo = construir_modelo()
arquitetura = pd.DataFrame([(c.name, c.__class__.__name__, str(c.output.shape), c.count_params()) for c in modelo.layers],
                           columns=["camada", "tipo", "saída", "parâmetros"])
display(arquitetura)
print(f"Parâmetros totais: {modelo.count_params():,} | entrada {modelo.input_shape} | saída {modelo.output_shape}")
````

````python
def treino_curto():
    m = construir_modelo()
    h = m.fit(X_tr[:1024], y_tr[:1024], batch_size=TAMANHO_LOTE, epochs=1, shuffle=True, verbose=0)
    return h.history["loss"][0], m.get_weights()


(perda_a, pesos_a), (perda_b, pesos_b) = treino_curto(), treino_curto()
diferenca = max(float(np.abs(a - b).max()) for a, b in zip(pesos_a, pesos_b))
print(f"Duas execuções de 1 época em 1.024 amostras com a mesma semente: perdas {perda_a:.6f} e {perda_b:.6f}; maior diferença nos pesos = {diferenca:.2e}")
````

### 5.3 Treinamento

Foram executadas **três tentativas**, todas com a mesma arquitetura e os mesmos dados, e a escolha seguiu uma regra fixa: a tentativa com menor perda de validação. As tentativas existem porque a primeira configuração apresentou instabilidade (veja a interpretação a seguir); as duas seguintes são correções mínimas e conhecidas para esse problema, e não uma busca de hiperparâmetros.

1. **Inicial:** Adam com taxa de 1e-3.
2. **Gradiente limitado:** a mesma taxa, com norma do gradiente limitada a 1,0 (`clipnorm`).
3. **Taxa menor:** Adam com taxa de 3e-4.

````python
class Cronometro(keras.callbacks.Callback):
    def on_train_begin(self, logs=None):
        self.duracoes, self.inicio = [], 0.0

    def on_epoch_begin(self, epoch, logs=None):
        self.inicio = time.perf_counter()

    def on_epoch_end(self, epoch, logs=None):
        self.duracoes.append(time.perf_counter() - self.inicio)


def treinar(**opcoes):
    """Treina só com X_tr/y_tr, acompanha X_va/y_va e restaura os pesos da melhor época (menor val_loss)."""
    m, cronometro = construir_modelo(**opcoes), Cronometro()
    parada = keras.callbacks.EarlyStopping(monitor="val_loss", patience=PACIENCIA, restore_best_weights=True)
    inicio = time.perf_counter()
    hist = m.fit(X_tr, y_tr, validation_data=(X_va, y_va), epochs=MAX_EPOCAS, batch_size=TAMANHO_LOTE, callbacks=[parada, cronometro], shuffle=True, verbose=0)
    tabela = pd.DataFrame(hist.history, index=pd.RangeIndex(1, len(hist.epoch) + 1, name="época"))
    tabela["segundos"] = cronometro.duracoes
    return {"modelo": m, "historico": hist, "tabela": tabela, "tempo": time.perf_counter() - inicio, "melhor_epoca": int(tabela["val_loss"].idxmin())}


TENTATIVAS = {"1. inicial (taxa 1e-3)": dict(taxa=1e-3),
              "2. gradiente limitado (taxa 1e-3, clipnorm 1,0)": dict(taxa=1e-3, clipnorm=1.0),
              "3. taxa menor (3e-4)": dict(taxa=3e-4)}
resultados = {nome: treinar(**opcoes) for nome, opcoes in TENTATIVAS.items()}

resumo = pd.DataFrame({nome: {"épocas executadas": len(r["tabela"]), "melhor época": r["melhor_epoca"], "menor val_loss": r["tabela"]["val_loss"].min(),
                              "val_accuracy na melhor época": r["tabela"].loc[r["melhor_epoca"], "val_accuracy"],
                              "maior aumento da perda de treino entre épocas": r["tabela"]["loss"].diff().max(), "tempo (s)": r["tempo"]}
                       for nome, r in resultados.items()}).T
display(resumo)
print("Acurácia de validação por época:")
for nome, r in resultados.items():
    print(f"  {nome}: {[round(v, 2) for v in r['tabela']['val_accuracy']]}")

escolhida = min(resultados, key=lambda nome: resultados[nome]["tabela"]["val_loss"].min())    # regra: menor perda de validação
modelo, historico, h, melhor_epoca = (resultados[escolhida][k] for k in ("modelo", "historico", "tabela", "melhor_epoca"))
print(f"Tentativa escolhida (menor val_loss): {escolhida} | tempo total das três tentativas: {sum(r['tempo'] for r in resultados.values()):.0f} s")
print(f"\nHistórico da tentativa escolhida (época {melhor_epoca} = melhor):")
h
````

**Interpretação.**

| Tentativa | Épocas | Melhor época | Menor perda de validação | Acurácia de validação | Tempo |
| :--- | ---: | ---: | ---: | ---: | ---: |
| 1. inicial (1e-3) | 4 | 1 | 0,53 | 0,81 | 188 s |
| 2. gradiente limitado (1e-3) | 5 | 2 | 0,39 | 0,86 | 186 s |
| 3. taxa menor (3e-4) | 9 | 6 | 0,37 | 0,87 | 268 s |

* **Problema encontrado:** a tentativa 1 foi **instável**. A perda de treino subiu entre épocas (aumento de 0,13), a acurácia de validação caiu de 0,81 (época 1) para 0,57 (época 2) e depois para 0,51 e 0,50, e o *early stopping* encerrou na época 4. Não se trata de *overfitting* (a perda de treino também piorou), mas de um problema de otimização, típico de LSTM com sequências longas.
* As tentativas 2 e 3 estabilizaram o treino. A **tentativa 3 teve a menor perda de validação (0,373)** e foi escolhida pela regra. A diferença para a tentativa 2 (0,39) é pequena.
* O tempo total das três tentativas foi de cerca de 642 s em CPU (12 núcleos, sem GPU); a tentativa escolhida levou 268 s, com 26 a 35 s por época. Os tempos dependem do ambiente.

3. **Limite de leitura:** os resultados são de validação em dados do mesmo gerador sintético, e não indicam capacidade de detectar fraude real. Teste e generalização, que medem outras coisas, ainda não foram avaliados.

**Limitações.**

* Uma única execução por tentativa (a variação entre sementes não foi medida) e apenas três tentativas, sem ajuste fino.
* A validação foi usada para *early stopping* e para escolher entre as tentativas, então as métricas de validação são ligeiramente otimistas.
* O limiar de 0,5 e a prevalência de 50% de anomalias foram assumidos; em uso real a proporção seria muito menor.
* O modelo só enxerga as primeiras 256 posições e foi treinado com anomalias nas primeiras 200 palavras.
* A reprodutibilidade foi verificada em CPU.

**Pendências.** Avaliar o modelo final no teste e na generalização, uma única vez, por tipo e por variante; investigar a repetição (por exemplo, agregando todos os estados da sequência em vez de usar só o último, ou com atenção, sempre decidindo com treino e validação); e só então passar à explicabilidade (LIME) e à interface (Gradio).

### 6.3 Desempenho por categoria e por mecanismo

Recall por tipo de anomalia nas três partições, ROC AUC por tipo (anomalias de um tipo contra as avaliações normais da mesma partição), recall por variante no teste e recall por mecanismo inédito na generalização, com intervalo de confiança de 95% (Wilson), porque os grupos da generalização são pequenos (31 a 58 avaliações).

````python
def wilson(k, n, z=1.96):
    if n == 0:
        return np.nan, np.nan
    p, d = k / n, 1 + z * z / n
    centro, margem = p + z * z / (2 * n), z * np.sqrt(p * (1 - p) / n + z * z / (4 * n * n))
    return max((centro - margem) / d, 0.0), min((centro + margem) / d, 1.0)


def recall_grupo(p, colunas):
    q = Q[p].assign(prob=prob[p], acerto=previsto[p] == 1)
    q = q[q["classe"] == 1]
    linhas = []
    for chave, g in q.groupby(colunas):
        lo, hi = wilson(int(g["acerto"].sum()), len(g))
        linhas.append({**dict(zip(colunas, chave if isinstance(chave, tuple) else (chave,))), "n": len(g), "recall": g["acerto"].mean(), "IC95 inf.": lo, "IC95 sup.": hi, "prob. média": g["prob"].mean()})
    return pd.DataFrame(linhas).set_index(colunas)


def auc_por_tipo(p):
    normais = Q[p]["classe"] == 0
    return {t: roc_auc_score(np.r_[np.zeros(normais.sum()), np.ones((Q[p]["tipo_anomalia"] == t).sum())],
                             np.r_[prob[p][normais], prob[p][Q[p]["tipo_anomalia"] == t]]) for t in TIPOS}


por_tipo6 = pd.concat({p: recall_grupo(p, ["tipo_anomalia"])["recall"] for p in PARTICOES}, axis=1)
por_tipo6.loc["falso positivo (classe 0)"] = [matrizes[p][0, 1] / matrizes[p][0].sum() for p in PARTICOES]
print("Recall por tipo de anomalia (e taxa de falsos positivos na última linha)")
display(por_tipo6)
print("ROC AUC por tipo (anomalias do tipo contra as avaliações normais da mesma partição)")
display(pd.DataFrame({p: auc_por_tipo(p) for p in PARTICOES}))

print("Teste: recall por variante (modelos do treino), com IC de 95%")
display(recall_grupo("teste", ["tipo_anomalia", "variante"]).sort_values("recall").head(8))
print("Generalização: recall por mecanismo inédito, com IC de 95%")
recall_grupo("generalizacao", ["tipo_anomalia", "variante"]).sort_values(["tipo_anomalia", "recall"])
````

**Interpretação.**

| Recall por tipo | Validação | Teste | Generalização | AUC no teste | AUC na generalização |
| :--- | ---: | ---: | ---: | ---: | ---: |
| Spam promocional | 0,99 | 0,99 | 0,13 | 0,98 | 0,58 |
| Manipulação de avaliações | 0,97 | 0,97 | 0,39 | 0,97 | 0,65 |
| Repetição excessiva | 0,41 | 0,44 | 0,09 | 0,70 | 0,51 |
| Falsos positivos (normais) | 0,05 | 0,04 | 0,06 | | |

* **Teste:** spam e manipulação são praticamente resolvidos (recall 0,97 a 0,99, AUC 0,97 a 0,98), enquanto a repetição fica em 0,44 (AUC 0,70). Dentro da repetição, as três formas ficam parecidas (frase 0,42, expressão 0,42, palavra 0,49).
* **Generalização, por tipo:** o spam cai de 0,99 para 0,13, a manipulação de 0,97 para 0,39 e a repetição de 0,44 para 0,09. O AUC da repetição (0,51) mostra que, nos mecanismos inéditos, o modelo não ordena as anomalias acima das normais.
* **Generalização, por mecanismo:** os melhores casos são parecidos com o vocabulário do treino: manipulação por ranking de revisores (0,69) e por revelação de pagamento (0,54), que ainda usam palavras como *helpful* e *rating*. Os piores são os mais distantes: manipulação em outros idiomas (0,00), repetição por alongamento de caracteres (0,00) e por sequência sem sentido (0,05), spam patrocinado, camuflado e multilíngue (0,08 a 0,11). Os intervalos são largos (por exemplo, 0,54 com IC de 0,40 a 0,67), então diferenças pequenas entre mecanismos não são conclusivas.

### 6.4 Análise de erros: repetição excessiva

A repetição é a categoria com pior recall no teste. A célula mede o recall por tamanho do trecho, por posição e **pela origem da unidade repetida** (repertório neutro do gerador ou texto da própria avaliação) e mostra exemplos reproduzíveis (semente 42) de repetições não detectadas e detectadas.

````python
t = Q["teste"].assign(prob=prob["teste"], acerto=previsto["teste"] == 1)
rep = t[t["tipo_anomalia"] == "repeticao_excessiva"].copy()
rep["palavras do trecho"] = rep["trecho_inserido"].str.count(" ") + 1
rep["faixa do trecho"] = pd.cut(rep["palavras do trecho"], [0, 15, 25, 40], labels=["até 15", "16 a 25", "26 a 40"])
rep["faixa de posição"] = pd.cut(rep["pos_palavra"], [-1, 20, 100, 160], labels=["0 a 20", "21 a 100", "101 a 160"])
print("Teste, repetição excessiva: recall por tamanho do trecho e por posição de inserção")
display(rep.groupby("faixa do trecho", observed=True)["acerto"].agg(["size", "mean"]).rename(columns={"size": "n", "mean": "recall"}))
display(rep.groupby("faixa de posição", observed=True)["acerto"].agg(["size", "mean"]).rename(columns={"size": "n", "mean": "recall"}))


def unidade(trecho):
    """Menor sequência de palavras que, repetida, forma o trecho."""
    tk = re.findall(r"[\w']+", trecho.lower())
    for n in range(1, len(tk) // 2 + 1):
        if len(tk) % n == 0 and tk == tk[:n] * (len(tk) // n):
            return " ".join(tk[:n])
    return " ".join(tk)


repertorio = {item for lista in POOL_REPETICAO.values() for item in lista}
rep["origem da unidade repetida"] = np.where(rep["trecho_inserido"].map(unidade).isin(repertorio), "repertório neutro do gerador", "texto da própria avaliação")
print("\nTeste, repetição excessiva: recall conforme a origem da unidade repetida")
display(rep.groupby("origem da unidade repetida")["acerto"].agg(["size", "mean"]).rename(columns={"size": "n", "mean": "recall"}))
display(rep.pivot_table(index="variante", columns="origem da unidade repetida", values="acerto", aggfunc="mean"))


def janela(linha, margem=90):
    texto, ini, fim = linha["texto_modelo"], linha["ini_anomalia"], linha["fim_anomalia"]
    return ("..." if ini > margem else "") + texto[max(0, ini - margem):ini] + "«" + texto[ini:fim] + "»" + texto[fim:fim + margem] + "..."


print("\nRepetições NÃO detectadas (falsos negativos), 2 por forma de repetição, amostra com semente fixa:")
for forma in ("palavra", "expressao", "frase"):
    for _, l in rep[(rep["variante"] == forma) & ~rep["acerto"]].sample(2, random_state=SEMENTE).iterrows():
        print(f"[{forma} | prob {l['prob']:.2f} | posição {l['pos_palavra']}] {janela(l)}")
print("\nRepetições detectadas (contraste), 1 por forma:")
for forma in ("palavra", "expressao", "frase"):
    l = rep[(rep["variante"] == forma) & rep["acerto"]].sample(1, random_state=SEMENTE).iloc[0]
    print(f"[{forma} | prob {l['prob']:.2f} | posição {l['pos_palavra']}] {janela(l)}")
````

**Interpretação.**

* **A origem da unidade repetida explica a maior parte do desempenho.** Quando a unidade vem do **repertório neutro do gerador** (por exemplo, "check this", "you know what i mean", "time"), o recall é de **0,74**. Quando vem do **texto da própria avaliação**, é de apenas **0,14**. O padrão se repete nas três formas (palavra 0,76 contra 0,17; expressão 0,70 contra 0,15; frase 0,76 contra 0,10).
* Os exemplos mostram o contraste: "really! really! ... really" (13 vezes) recebe probabilidade 0,15, e "disavowed. disavowed. ..." recebe 0,14, ao passo que "check this check this ..." recebe 0,97. A repetição de uma palavra qualquer, embora muito marcada, não é percebida.
* **Interpretação:** isso indica que o modelo **aprendeu o vocabulário das unidades do repertório do gerador, e não o padrão de repetição**. É um sinal claro de memorização de templates, e coincide com a queda observada na generalização.
* O tamanho do trecho também importa: o recall é de 0,43 para trechos de até 15 palavras, 0,51 para 16 a 25 e apenas 0,14 para 26 a 40 palavras. A posição de inserção tem efeito pequeno (0,50 no início, 0,44 e 0,43 adiante). Como o repertório neutro e o tamanho do trecho se misturam, não é possível separar as causas apenas com estes números.

### 7.1 Modelo, pré-processamento e `predict_proba`

O LIME precisa de uma função que receba uma **lista de textos brutos** e devolva as **probabilidades das duas classes**, uma linha por texto. `predict_proba` faz isso: vetoriza os textos com a mesma camada da Etapa 04 (reconstruída a partir do vocabulário salvo), aplica o modelo e devolve `[P(normal), P(anomalia sintética)]`, com `P(normal) = 1 - P(classe 1)`.

````python
import html
import importlib.util
import json
import os
import subprocess
import sys

import tensorflow as tf
from IPython.display import HTML, display

for pacote in ("lime", "gradio"):                       # só instala se ainda não estiver disponível
    if importlib.util.find_spec(pacote) is None:
        subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", pacote])
from lime.lime_text import LimeTextExplainer

PASTA = next(p for p in ("artefatos", "/content/artefatos") if os.path.isdir(p))
modelo_final = keras.saving.load_model(f"{PASTA}/lstm_melhor_val_loss.keras")           # o mesmo modelo da Etapa 05, sem retreinar
dados_voc = json.load(open(f"{PASTA}/vocabulario.json", encoding="utf-8"))
vocabulario_final, COMPRIMENTO = dados_voc["vocabulario"], dados_voc["comprimento"]
if "padronizar" not in globals():                         # função de pré-processamento da Etapa 04, salva em artefatos/padronizar.py
    exec(open(f"{PASTA}/padronizar.py", encoding="utf-8").read(), globals())
camada_final = keras.layers.TextVectorization(max_tokens=len(vocabulario_final), standardize=padronizar, split="whitespace",
                                              output_mode="int", output_sequence_length=COMPRIMENTO)
camada_final.set_vocabulary(vocabulario_final[2:])        # vocabulário do treino (índices 0 e 1 são padding e desconhecido)
pesos_antes = hash(b"".join(w.tobytes() for w in modelo_final.get_weights()))


def vetorizar_final(textos, lote=1024):
    textos = [textos] if isinstance(textos, str) else list(textos)
    return np.concatenate([camada_final(tf.constant(textos[i:i + lote])).numpy() for i in range(0, len(textos), lote)]).astype("int32")


def predict_proba(textos):
    """Lista de textos brutos -> matriz (n, 2) com [P(normal), P(anomalia sintética)], o formato esperado pelo LIME."""
    textos = [textos] if isinstance(textos, str) else list(textos)
    p1 = modelo_final.predict(vetorizar_final(textos), batch_size=512, verbose=0).ravel().astype("float64")
    return np.column_stack([1 - p1, p1])


registro = {"chamadas": 0, "textos": 0, "minimo": 1.0, "maximo": 0.0, "soma_maxima_desvio": 0.0}


def predict_proba_registrado(textos):
    """Mesma função, anotando o que o LIME envia e recebe (para validar as probabilidades)."""
    p = predict_proba(textos)
    registro["chamadas"] += 1
    registro["textos"] += len(p)
    registro["minimo"], registro["maximo"] = min(registro["minimo"], p.min()), max(registro["maximo"], p.max())
    registro["soma_maxima_desvio"] = max(registro["soma_maxima_desvio"], float(np.abs(p.sum(axis=1) - 1).max()))
    return p


npz = np.load(f"{PASTA}/sequencias.npz")
amostra_val = dataset[dataset["particao"] == "validacao"].reset_index(drop=True).iloc[:300]
mesma_sequencia = bool((vetorizar_final(list(amostra_val["texto_modelo"])) == npz["X_validacao"][:300]).all())
print(f"Modelo: {modelo_final.count_params():,} parâmetros | vocabulário {len(vocabulario_final):,} | comprimento {COMPRIMENTO}")
print("Pré-processamento reconstruído a partir dos artefatos gera as mesmas sequências salvas (300 textos de validação):", mesma_sequencia)
print("predict_proba(['great movie', '']) ->\n", predict_proba(["great movie", ""]).round(4))
````

### 7.3 Explicações LIME

Para cada exemplo, o LIME remove palavras aleatoriamente do texto (1.000 variações), pergunta ao modelo a probabilidade da classe 1 em cada variação e ajusta um modelo linear simples que atribui um **peso a cada palavra**. Peso positivo (vermelho) favorece a classe "anomalia"; peso negativo (azul) favorece "normal". O explicador é criado a cada chamada com semente fixa, de modo que o mesmo texto sempre gera a mesma explicação. Cada exemplo mostra o gráfico de pesos e o texto com as palavras destacadas.

````python
def explicar(texto, num_samples=1000, num_features=12):
    """Explicação LIME da classe 1. O explicador é criado a cada chamada, com semente fixa, para que o mesmo texto dê sempre a mesma explicação."""
    explicador = LimeTextExplainer(class_names=["normal", "anomalia"], random_state=SEMENTE)
    return explicador.explain_instance(texto, predict_proba_registrado, num_features=num_features, num_samples=num_samples, labels=(1,))


def valor_lime(x):
    return float(np.ravel(x[1] if isinstance(x, dict) else x)[0])


def destacar_html(texto, pesos):
    """Texto com as palavras coloridas: vermelho favorece 'anomalia'; azul favorece 'normal'. A intensidade é proporcional ao peso."""
    maximo = max((abs(w) for w in pesos.values()), default=0) or 1

    def pintar(m):
        palavra, peso = m.group(0), pesos.get(m.group(0))
        if peso is None:
            return html.escape(palavra)
        intensidade = 0.20 + 0.65 * abs(peso) / maximo
        cor = f"rgba(214,39,40,{intensidade:.2f})" if peso > 0 else f"rgba(31,119,180,{intensidade:.2f})"
        return f'<span style="background:{cor};border-radius:3px;padding:0 2px" title="peso LIME {peso:+.3f}">{html.escape(palavra)}</span>'
    corpo = re.sub(r"\w+|\W+", pintar, texto)
    legenda = '<div style="font-size:12px;margin-bottom:6px"><span style="background:rgba(214,39,40,.6);padding:0 6px">favorece anomalia</span> <span style="background:rgba(31,119,180,.6);padding:0 6px">favorece normal</span> (apenas as palavras mais influentes são coloridas)</div>'
    return f'{legenda}<div style="white-space:pre-wrap;line-height:1.7;font-family:sans-serif">{corpo}</div>'


def grafico_lime(exp, titulo):
    palavras, pesos = zip(*sorted(exp.as_list(label=1), key=lambda x: x[1]))
    fig, ax = plt.subplots(figsize=(7, 0.32 * len(palavras) + 1.3))
    ax.barh(palavras, pesos, color=["#d62728" if w > 0 else "#1f77b4" for w in pesos])
    ax.axvline(0, color="k", linewidth=0.8)
    ax.set_title(titulo, fontsize=10)
    ax.set_xlabel("Peso LIME (positivo favorece anomalia; negativo favorece normal)")
    ax.set_ylabel("Palavra")
    plt.tight_layout()
    plt.show()


explicacoes, resumo = {}, {}
for nome, e in exemplos.items():
    exp = explicar(e["texto_modelo"])
    explicacoes[nome] = exp
    resumo[nome] = {"prob. do modelo": e["prob"], "previsão local do LIME": valor_lime(exp.local_pred), "R² local": valor_lime(exp.score)}
    print(f"\n=== {nome} | classe real {int(e['classe'])} | probabilidade da classe 1 = {e['prob']:.3f}")
    print("Palavras mais influentes:", ", ".join(f"{w} ({p:+.2f})" for w, p in sorted(exp.as_list(label=1), key=lambda x: -abs(x[1]))[:6]))
    grafico_lime(exp, nome)
    display(HTML(destacar_html(e["texto_modelo"], dict(exp.as_list(label=1)))))
print("\nFidelidade do modelo local do LIME:")
pd.DataFrame(resumo).T
````

### 8.1 Função de análise

A função `analisar(texto)` concentra a lógica da interface: trata entrada vazia sem chamar o modelo, conta os tokens do texto com a mesma padronização, calcula a probabilidade da classe 1 com `predict_proba`, decide com o limiar de 0,5 (**Normal** ou **Anomalia sintética**), gera a explicação LIME (800 variações) e avisa quando o texto passa de 256 tokens. A célula a seguir a testa diretamente com quatro entradas: normal, suspeita, vazia e longa.

````python
import gradio as gr

LIMITE_TOKENS = COMPRIMENTO
AVISO = f"""**Limitações.** (1) Apenas os **primeiros {LIMITE_TOKENS} tokens** do texto são analisados; o restante é ignorado. (2) O modelo foi treinado com anomalias **sintéticas** (spam promocional, repetição excessiva e manipulação de avaliações) e **não detecta fraude real**. (3) Ele reconhece bem os modelos vistos no treino, mas tem **baixa generalização** para mecanismos inéditos (recall de 0,20 na avaliação final) e falha em repetições feitas com palavras do próprio texto. (4) A explicação do LIME é **local e aproximada**: indica o efeito de cada palavra neste texto, não uma regra do modelo. (5) Use avaliações em inglês."""


def contar_tokens(texto):
    return len(padronizar(tf.constant([texto])).numpy()[0].decode("utf-8").split())


def analisar(texto):
    """Devolve (classificação, probabilidade da classe 1, explicação em HTML, tabela de palavras, aviso desta análise)."""
    texto = (texto or "").strip()
    if not texto:
        return "Insira uma avaliação para analisar.", None, "", [], ""
    n_tokens = contar_tokens(texto)
    p1 = float(predict_proba([texto])[0, 1])
    rotulo = "Anomalia sintética" if p1 >= 0.5 else "Normal"
    aviso = f"O texto tem {n_tokens} tokens: apenas os primeiros {LIMITE_TOKENS} foram analisados pelo modelo." if n_tokens > LIMITE_TOKENS else ""
    try:
        exp = explicar(texto, num_samples=800)
        pares = exp.as_list(label=1)
        visual = destacar_html(texto, dict(pares))
        tabela = [[palavra, round(peso, 4), "anomalia" if peso > 0 else "normal"] for palavra, peso in sorted(pares, key=lambda x: -abs(x[1]))]
    except Exception as erro:                                              # por exemplo, texto sem palavras para explicar
        visual, tabela, aviso = "", [], (aviso + " Explicação LIME indisponível para este texto.").strip()
    return rotulo, p1, visual, tabela, aviso


casos = {"normal": exemplos["2. acerto: avaliação normal (teste)"]["texto_modelo"], "suspeita (spam)": exemplos["1. acerto: spam detectado (teste)"]["texto_modelo"],
         "vazia": "   ", "longa": QL["teste"].loc[QL["teste"]["n_palavras_texto"] > 450, "texto_modelo"].iloc[0]}
linhas = {}
for nome, texto in casos.items():
    rotulo, p1, visual, tabela, aviso = analisar(texto)
    linhas[nome] = {"tokens": contar_tokens(texto) if texto.strip() else 0, "classificação": rotulo, "probabilidade": p1, "palavras na tabela": len(tabela), "aviso": aviso[:60]}
print("Teste direto da função `analisar` (a mesma usada pela interface):")
pd.DataFrame(linhas).T
````

### 8.2 Interface Gradio

A interface tem um campo de texto, o botão **Analisar**, a classificação, a probabilidade da classe 1, o texto com as palavras destacadas pelo LIME, a tabela das palavras mais influentes e um aviso fixo sobre os limites (256 tokens, anomalias sintéticas, baixa generalização, LIME aproximado). Três exemplos prontos podem ser carregados com um clique. A interface usa exatamente `analisar`, portanto o mesmo modelo e o mesmo pré-processamento.

A célula inicia o servidor sem bloquear o notebook. Para abri-la em uma nova sessão, execute esta célula; em ambientes como o Colab ela aparece abaixo da célula. Para encerrá-la, use `demo.close()`.

````python
EXEMPLOS = [[casos["normal"]], [casos["suspeita (spam)"]], ["This movie is great great great great great great great great great great great great, one of the best I have seen."]]

with gr.Blocks(title="Detector de anomalias sintéticas em avaliações") as demo:
    gr.Markdown("# Anomalias sintéticas em avaliações de filmes\nCole uma avaliação (em inglês) e clique em **Analisar**.")
    gr.Markdown(AVISO)
    entrada = gr.Textbox(label="Avaliação", lines=8, placeholder="Cole aqui o texto da avaliação...")
    botao = gr.Button("Analisar", variant="primary")
    with gr.Row():
        classificacao = gr.Textbox(label="Classificação")
        probabilidade = gr.Number(label="Probabilidade da classe 1 (anomalia sintética)")
    aviso_analise = gr.Markdown()
    explicacao = gr.HTML(label="Explicação visual (LIME)")
    tabela_palavras = gr.Dataframe(headers=["palavra", "peso LIME", "favorece"], label="Palavras mais influentes (LIME)", interactive=False)
    botao.click(analisar, inputs=entrada, outputs=[classificacao, probabilidade, explicacao, tabela_palavras, aviso_analise], api_name="analisar")
    gr.Examples(examples=EXEMPLOS, inputs=entrada, label="Exemplos")

_, URL_LOCAL, _ = demo.launch(share=False, prevent_thread_lock=True, quiet=True)     # inicia o servidor sem bloquear o notebook
print("Interface iniciada em", URL_LOCAL)
````
